1. Define Core KPIs

KPI 1: Total Sales

Formula:
Total Sales = Sum of Sales Column

Business Rationale:
Measures overall revenue generated from all transactions and indicates business performance.

-------------------------------------------------

KPI 2: Average Sales Value

Formula:
Average Sales = Total Sales / Number of Transactions

Business Rationale:
Shows average revenue generated per transaction.

-------------------------------------------------

KPI 3: Top Product Performance

Formula:
Product Performance = Sum of Sales grouped by Product

Business Rationale:
Identifies highest revenue generating product and helps business prioritize inventory and marketing.

-------------------------------------------------

KPI 4: Regional Sales Performance

Formula:
Regional Sales = Sum of Sales grouped by Region

Business Rationale:
Helps identify top performing sales regions for business growth strategies.

-------------------------------------------------

KPI 5: Transaction Count

Formula:
Transaction Count = Total Number of Records

Business Rationale:
Measures business activity volume and customer purchase frequency.

```python
import pandas as pd


df = pd.read_excel(r"C:\Users\DELL\Downloads\sales.data\Train.xlsx")

df.head()
```

2. Deep-Dive Analysis

Selected Business Area: Segmentation Analysis

Objective:
Analyze sales performance across different products and regions to identify business segments and patterns.

Segmentation Rules:
1. Product-based segmentation
2. Region-based segmentation

Business Goal:
Identify high-performing products and regions to improve business decisions and sales strategies.

```python
# KPI 1
print("Total Sales:", df['TotalSalesValue'].sum())

# KPI 2
print("Average Sales:", df['TotalSalesValue'].mean())

# KPI 3
print(df.groupby('ProductID')['TotalSalesValue'].sum())

# KPI 4
print(df.groupby('SalesPersonID')['TotalSalesValue'].sum())

# KPI 5
print("Transactions:", len(df))
```

```python
# Product segmentation

product_sales = df.groupby('ProductID')['TotalSalesValue'].sum()

print("Product Segmentation")
print(product_sales)

# SalesPerson segmentation

salesperson_sales = df.groupby('SalesPersonID')['TotalSalesValue'].sum()

print("\nSalesPerson Segmentation")
print(salesperson_sales)
```

```python
import matplotlib.pyplot as plt

# Product segmentation chart

product_sales.plot(
    kind='bar',
    figsize=(5,4),
    title='Sales by Product'
)

plt.ylabel("Total Sales Value")
plt.show()


# SalesPerson segmentation chart

salesperson_sales.plot(
    kind='bar',
    figsize=(5,4),
    title='Sales by SalesPerson'
)

plt.ylabel("Total Sales Value")
plt.show()
```

Conclusion:

Laptop generated highest revenue.

West region showed highest sales performance.

Product and regional segmentation help businesses prioritize resources and improve decision-making.
